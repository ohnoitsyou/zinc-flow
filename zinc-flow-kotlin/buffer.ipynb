```kotlin
import org.apache.commons.collections4.queue.CircularFifoQueue

data class Event(val id: Int)

val buffer = CircularFifoQueue<Event>(25)

(0 until 30).forEach {
    buffer.add(Event(it))
}

val last15 = buffer.reversed().take(15).reversed()
(30 until 60).forEach {
    buffer.add(Event(it))
}
val last10 = buffer.asIterable().reversed().take(10).reversed()
println("last 15: $last15")
println("last 10: $last10")
println("same size: ${buffer.reversed().take(30).reversed()}")
println("buffer: $buffer; size: ${buffer.size}")
```
